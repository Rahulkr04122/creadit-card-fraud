```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✓ All libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv('../data/raw/creditcard.csv')

# Display basic information
print("Dataset Shape:", df.shape)
print("\n" + "="*50)
print("First 5 rows:")
print(df.head())
```

```python
# Statistical summary
print("\nBasic Statistics:")
print(df.describe())
```

```python
# Check the target variable (Class: 0=Legitimate, 1=Fraud)
print("\nTarget Variable Distribution:")
print(df['Class'].value_counts())
print("\nProportions:")
print(df['Class'].value_counts(normalize=True))
```

```python
# Check all column names
print("All Features:")
for i, col in enumerate(df.columns):
    print(f"{i}: {col}")
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Visualize class distribution
plt.figure(figsize=(8, 5))
df['Class'].value_counts().plot(kind='bar', color=['green', 'red'])
plt.title('Transaction Class Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Class (0=Legitimate, 1=Fraud)', fontsize=8, fontweight='bold')
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("Green bar = Legitimate (99.83%)")
print("Red bar = Fraud (0.17%)")
```

```python
# Compare statistics: Fraud vs Legitimate
print("LEGITIMATE TRANSACTIONS (Class = 0):")
print(df[df['Class'] == 0]['Amount'].describe())

print("\n" + "="*50)
print("FRAUDULENT TRANSACTIONS (Class = 1):")
print(df[df['Class'] == 1]['Amount'].describe())
```

```python
# Time patterns
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# All transactions over time
axes[0].scatter(df[df['Class'] == 0]['Time'], 
                df[df['Class'] == 0]['Amount'], 
                alpha=0.5, s=10, color='green', label='Legitimate')
axes[0].scatter(df[df['Class'] == 1]['Time'], 
                df[df['Class'] == 1]['Amount'], 
                alpha=0.8, s=20, color='red', label='Fraud')
axes[0].set_title('All Transactions Over Time', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Time (seconds from start)', fontsize=11)
axes[0].set_ylabel('Amount ($)', fontsize=11)
axes[0].legend()
axes[0].grid(alpha=0.3)

# Fraud count over time
fraud_per_hour = df[df['Class'] == 1].groupby(df['Time'] // 3600).size()
axes[1].plot(fraud_per_hour.index, fraud_per_hour.values, marker='o', color='red', linewidth=2)
axes[1].set_title('Fraudulent Transactions Over Time', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Time (hours from start)', fontsize=11)
axes[1].set_ylabel('Fraud Count', fontsize=11)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Calculate correlation with target
correlation_with_target = df.corr()['Class'].sort_values(ascending=False)

print("Features Most Correlated with Fraud:")
print(correlation_with_target.head(10))
print("\n" + "="*50)
print("Features Least Correlated with Fraud:")
print(correlation_with_target.tail(10))
```

```python
# Correlation heatmap (top features only)
plt.figure(figsize=(12, 8))

# Select top correlated features + target
top_features = correlation_with_target.abs().nlargest(15).index.tolist()
correlation_matrix = df[top_features].corr()

sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Matrix - Top 15 Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Features most correlated with fraud (by absolute value):")
print(correlation_with_target.abs().nlargest(10))
```

```python
# Box plots for key features
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

# Select top 6 features (excluding Class and Time)
top_6_features = correlation_with_target.abs().nlargest(7).index.tolist()
top_6_features = [f for f in top_6_features if f != 'Class'][:6]

for idx, feature in enumerate(top_6_features):
    data_to_plot = [df[df['Class'] == 0][feature], 
                    df[df['Class'] == 1][feature]]
    
    bp = axes[idx].boxplot(data_to_plot)
    axes[idx].set_xticklabels(['Legitimate', 'Fraud'])
    axes[idx].set_title(f'{feature} Distribution', fontweight='bold')
    axes[idx].set_ylabel('Value')
    axes[idx].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("Box plots show outliers (dots above/below the box)")
```

```python
# Check for duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
print("Number of rows before removing duplicates:", df.shape[0])

# Remove duplicates if any
df_clean = df.drop_duplicates()

print("Number of rows after removing duplicates:", df_clean.shape[0])
```

```python
# Check missing values in detail
print("Missing values per column:")
print(df_clean.isnull().sum())

print("\nTotal missing values:", df_clean.isnull().sum().sum())
print("Percentage missing:", (df_clean.isnull().sum().sum() / (df_clean.shape[0] * df_clean.shape[1])) * 100)
```

```python
# Check for outliers using statistical method
# Outlier = value that is more than 3 standard deviations from mean

print("Checking for outliers in Amount column:")
print("\nAmount statistics:")
print(df_clean['Amount'].describe())

# Calculate z-score for Amount
from scipy import stats
z_scores = np.abs(stats.zscore(df_clean['Amount']))
outliers_count = (z_scores > 3).sum()

print(f"\nOutliers in Amount (> 3 std dev): {outliers_count}")
print(f"Percentage of outliers: {(outliers_count / len(df_clean)) * 100:.2f}%")
```

```python
# Important insight about outliers in fraud detection
print("Fraud characteristics:")
fraud_data = df_clean[df_clean['Class'] == 1]
legitimate_data = df_clean[df_clean['Class'] == 0]

print(f"\nAverage fraud amount: ${fraud_data['Amount'].mean():.2f}")
print(f"Average legitimate amount: ${legitimate_data['Amount'].mean():.2f}")

print(f"\nMax fraud amount: ${fraud_data['Amount'].max():.2f}")
print(f"Max legitimate amount: ${legitimate_data['Amount'].max():.2f}")

print("\n⚠️ IMPORTANT:")
print("Some large transactions ARE fraud, some are legitimate.")
print("We CANNOT simply remove outliers based on Amount.")
print("Removing them would delete important fraud examples.")
```

```python
# Compare the scale of different features
print("Feature ranges (min to max):")
print("\nAmount column:")
print(f"  Min: {df_clean['Amount'].min()}")
print(f"  Max: {df_clean['Amount'].max()}")
print(f"  Range: {df_clean['Amount'].max() - df_clean['Amount'].min()}")

print("\nV1 column:")
print(f"  Min: {df_clean['V1'].min():.2f}")
print(f"  Max: {df_clean['V1'].max():.2f}")
print(f"  Range: {df_clean['V1'].max() - df_clean['V1'].min():.2f}")

print("\n⚠️ PROBLEM:")
print("Amount ranges from 0 to 25,691")
print("V1 ranges from -56.41 to 2.45")
print("Models get confused by different scales!")
```

```python
# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Separate features (X) and target (y)
X = df_clean.drop('Class', axis=1)  # All columns except Class
y = df_clean['Class']                 # Only the Class column

print("Shape of features (X):", X.shape)
print("Shape of target (y):", y.shape)

# Scale features using StandardScaler
# StandardScaler: transforms data to mean=0, std=1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier handling
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print("\n✓ Features scaled successfully")
print("\nBefore scaling - Amount stats:")
print(X['Amount'].describe())
print("\nAfter scaling - Amount stats:")
print(X_scaled['Amount'].describe())
```

```python
from sklearn.model_selection import train_test_split

# Split into train (70%) and test (30%)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, 
    test_size=0.3,           # 30% for testing
    random_state=42,         # For reproducibility
    stratify=y               # Keep same class ratio in train and test
)

print("Training set size:", X_train.shape[0])
print("Test set size:", X_test.shape[0])

print("\nClass distribution in training set:")
print(y_train.value_counts())
print("\nClass distribution in test set:")
print(y_test.value_counts())

print("\n✓ Train/Test split complete")
print("We'll train on 70%, evaluate on 30% (unseen data)")
```

```python
# Save processed data for later use
import os

# Create output path if it doesn't exist
os.makedirs('../data/processed', exist_ok=True)

# Save datasets
pd.DataFrame(X_train, columns=X.columns).to_csv('../data/processed/X_train.csv', index=False)
pd.DataFrame(X_test, columns=X.columns).to_csv('../data/processed/X_test.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)

print("✓ Processed data saved to data/processed/")
print("  - X_train.csv (training features)")
print("  - X_test.csv (test features)")
print("  - y_train.csv (training target)")
print("  - y_test.csv (test target)")
```

```python
# Save the scaler for future use
import joblib

joblib.dump(scaler, '../models/scaler.pkl')

print("✓ Scaler saved to models/scaler.pkl")
print("Later, we'll load this scaler to scale new predictions")
```

```python
# Load the processed data we saved
X_train = pd.read_csv('../data/processed/X_train.csv')
X_test = pd.read_csv('../data/processed/X_test.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').values.ravel()
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()

print("Data loaded successfully!")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression

# Create the model
print("Creating Logistic Regression model...")
model_lr = LogisticRegression(random_state=42, max_iter=1000)

# Train the model
print("Training on", X_train.shape[0], "transactions...")
model_lr.fit(X_train, y_train)

print("✓ Model trained successfully!")
print(f"\nModel learned weights for {len(model_lr.coef_[0])} features")
```

```python
# Make predictions on test data
print("Making predictions on test data...")
y_pred = model_lr.predict(X_test)
y_pred_proba = model_lr.predict_proba(X_test)

print("✓ Predictions made!")

print(f"\nFirst 10 predictions:")
print(y_pred[:10])

print(f"\nFirst 10 prediction probabilities:")
print(y_pred_proba[:10])

print("\nExplanation:")
print("- Prediction: 0 or 1 (the class we predict)")
print("- Probability: [P(class=0), P(class=1)]")
print("- Example: [0.98, 0.02] means 98% legitimate, 2% fraud")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("LOGISTIC REGRESSION - EVALUATION METRICS")
print("=" * 50)
print(f"Accuracy:  {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall:    {recall:.4f} ({recall*100:.2f}%)")
print(f"F1-Score:  {f1:.4f}")
```

```python
# Confusion Matrix to understand the metrics
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
print("CONFUSION MATRIX:")
print("=" * 50)
print("\n                Predicted")
print("              Legitimate  Fraud")
print(f"Actual Legit    {cm[0][0]:<10}  {cm[0][1]:<10}")
print(f"       Fraud    {cm[1][0]:<10}  {cm[1][1]:<10}")

# Extract values
tn = cm[0][0]  # True Negatives (correctly predicted legitimate)
fp = cm[0][1]  # False Positives (predicted fraud, actually legitimate)
fn = cm[1][0]  # False Negatives (predicted legitimate, actually fraud)
tp = cm[1][1]  # True Positives (correctly predicted fraud)

print("\n" + "=" * 50)
print(f"True Negatives (TN):   {tn} (correctly identified legitimate)")
print(f"False Positives (FP):  {fp} (legitimate rejected as fraud)")
print(f"False Negatives (FN):  {fn} (fraud slipped through)")
print(f"True Positives (TP):   {tp} (correctly identified fraud)")

print("\n" + "=" * 50)
print("METRIC FORMULAS:")
print(f"\nAccuracy = (TP + TN) / Total")
print(f"         = ({tp} + {tn}) / {tp+tn+fp+fn}")
print(f"         = {accuracy:.4f}")

print(f"\nPrecision = TP / (TP + FP)")
print(f"          = {tp} / ({tp} + {fp})")
print(f"          = {precision:.4f}")
print("          = 'Of predictions we made as fraud, how many were correct?'")

print(f"\nRecall = TP / (TP + FN)")
print(f"       = {tp} / ({tp} + {fn})")
print(f"       = {recall:.4f}")
print("       = 'Of actual frauds, how many did we catch?'")

print(f"\nF1-Score = 2 * (Precision * Recall) / (Precision + Recall)")
print(f"         = {f1:.4f}")
print("         = Harmonic mean of precision and recall")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("TRAINING ALL 4 MODELS")
print("=" * 80)

# MODEL 1: Logistic Regression
print("\n1. Logistic Regression... ", end='', flush=True)
lr_model = LogisticRegression(random_state=42, max_iter=1000)
lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)
lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_precision = precision_score(y_test, y_pred_lr)
lr_recall = recall_score(y_test, y_pred_lr)
lr_f1 = f1_score(y_test, y_pred_lr)
print("✓")

# MODEL 2: Decision Tree
print("2. Decision Tree... ", end='', flush=True)
dt_model = DecisionTreeClassifier(random_state=42, max_depth=10)
dt_model.fit(X_train, y_train)
y_pred_dt = dt_model.predict(X_test)
dt_accuracy = accuracy_score(y_test, y_pred_dt)
dt_precision = precision_score(y_test, y_pred_dt)
dt_recall = recall_score(y_test, y_pred_dt)
dt_f1 = f1_score(y_test, y_pred_dt)
print("✓")

# MODEL 3: Random Forest
print("3. Random Forest... ", end='', flush=True)
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=15, n_jobs=-1)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)
print("✓")

# MODEL 4: KNN
print("4. KNN... ", end='', flush=True)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
y_pred_knn = knn_model.predict(X_test)
knn_accuracy = accuracy_score(y_test, y_pred_knn)
knn_precision = precision_score(y_test, y_pred_knn)
knn_recall = recall_score(y_test, y_pred_knn)
knn_f1 = f1_score(y_test, y_pred_knn)
print("✓")

print("\n" + "=" * 80)
print("✓ ALL 4 MODELS TRAINED SUCCESSFULLY")
print("=" * 80)

# Print individual results
print("\nLOGISTIC REGRESSION:")
print(f"  Accuracy: {lr_accuracy:.4f}, Precision: {lr_precision:.4f}, Recall: {lr_recall:.4f}, F1: {lr_f1:.4f}")

print("\nDECISION TREE:")
print(f"  Accuracy: {dt_accuracy:.4f}, Precision: {dt_precision:.4f}, Recall: {dt_recall:.4f}, F1: {dt_f1:.4f}")

print("\nRANDOM FOREST:")
print(f"  Accuracy: {rf_accuracy:.4f}, Precision: {rf_precision:.4f}, Recall: {rf_recall:.4f}, F1: {rf_f1:.4f}")

print("\nKNN:")
print(f"  Accuracy: {knn_accuracy:.4f}, Precision: {knn_precision:.4f}, Recall: {knn_recall:.4f}, F1: {knn_f1:.4f}")
```
